```python
from collections import Counter

def count_distinct_question_ids(txt_path: str, topk: int = 20):
    with open(txt_path, "r", encoding="utf-8") as f:
        lines = [ln.strip() for ln in f if ln.strip()]

    step = 6  # 常见 write_txt：每个用户 6 行
    q_counter = Counter()
    blocks = 0

    for i in range(0, len(lines), step):
        if i + 1 >= len(lines):
            break

        problems_line = lines[i + 1]  # 第2行：question_id 序列（逗号分隔）
        qids = [x.strip() for x in problems_line.split(",") if x.strip()]

        for qid in qids:
            if qid != "NA":
                q_counter[qid] += 1

        blocks += 1

    print("===== question_id stats =====")
    print(f"users(blocks parsed): {blocks}")
    print(f"total question_id tokens: {sum(q_counter.values())}")
    print(f"distinct question_id: {len(q_counter)}")

    print(f"\nTop {topk} most frequent question_id:")
    for q, c in q_counter.most_common(topk):
        print(f"{c}\t{q}")

if __name__ == "__main__":
    txt_path = r"data.txt"   # 改成你的真实路径
    count_distinct_question_ids(txt_path, topk=30)
```

```python
import json

def extract_difficulty_map(json_file_path, output_file_path=None):
    with open(json_file_path, "r", encoding="utf-8") as f:
        data = json.load(f)

    result = {}

    for item in data:
        song_id = item.get("song_id")
        exercise_id = item.get("exercise_id")
        difficulty_level = item.get("difficulty_level")

        if song_id is not None and exercise_id is not None and difficulty_level is not None:
            key = f"{song_id}^{exercise_id}"
            result[key] = difficulty_level

    if output_file_path:
        with open(output_file_path, "w", encoding="utf-8") as f:
            json.dump(result, f, ensure_ascii=False, indent=2)

    return result


if __name__ == "__main__":
    # json_file_path = "yousician_ukulele.json"
    json_file_path = r"C:\Users\chill\OneDrive - Australian National University\Course\25s2\research\datasets\chunked_ukelele\yousician_ukulele_original.json"
    output_file_path = "difficulty_map.json"

    difficulty_map = extract_difficulty_map(json_file_path, output_file_path)
    print("提取完成：")
    print(difficulty_map)
```

```python
import json


def extract_song_info(json_file_path, output_file_path=None):
    with open(json_file_path, "r", encoding="utf-8") as f:
        data = json.load(f)

    result = {}

    for item in data:
        song_id = item.get("song_id")
        exercise_id = item.get("exercise_id")
        difficulty_level = item.get("difficulty_level")
        events_data_raw = item.get("events_data")

        if song_id is None or exercise_id is None:
            continue

        key = f"{song_id}_{exercise_id}"

        pitches = []
        duration = []
        strings = []

        if events_data_raw:
            try:
                events_data = json.loads(events_data_raw)
                inner_data = events_data.get("data", {})

                pitches = inner_data.get("pitches") or []
                duration = inner_data.get("duration") or []
                strings = inner_data.get("strings") or []

            except (json.JSONDecodeError, TypeError):
                print(f"Warning: 无法解析 events_data, key={key}")
                continue

        entry = {
            "difficulty_level": difficulty_level,
            "pitches": pitches,
            "duration": duration,
            "strings": strings,
        }

        new_pitch_len = len(pitches)

        if key not in result:
            result[key] = entry
        else:
            old_pitch_len = len(result[key]["duration"]) if result[key]["duration"] is not None else 0

            # 只按 pitches 长度决定是否替换整条记录
            if new_pitch_len > old_pitch_len:
                result[key] = entry

            # 如果 pitches 长度一样，可以加一个次级规则
            # 例如优先保留 duration 更长的那条
            # elif new_pitch_len == old_pitch_len:
            #     old_duration_len = len(result[key]["duration"]) if result[key]["duration"] is not None else 0
            #     new_duration_len = len(duration)

            #     old_strings_len = len(result[key]["strings"]) if result[key]["strings"] is not None else 0
            #     new_strings_len = len(strings)

            #     if (new_duration_len + new_strings_len) > (old_duration_len + old_strings_len):
            #         result[key] = entry

    if output_file_path:
        with open(output_file_path, "w", encoding="utf-8") as f:
            json.dump(result, f, ensure_ascii=False, indent=2)

    return result


if __name__ == "__main__":
    json_file_path = r"C:\Users\chill\OneDrive - Australian National University\Course\25s2\research\datasets\chunked_ukelele\yousician_ukulele_original.json"
    output_file_path = "song_info_map.json"

    result = extract_song_info(json_file_path, output_file_path)
    print(json.dumps(result, ensure_ascii=False, indent=2))
```
